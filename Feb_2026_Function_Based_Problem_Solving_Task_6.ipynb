# **Problem Statement 1:Smart Parking Lot Management System**
## **Design a function to manage a smart parking lot.**
## The system should:
* Accept vehicle entry and exit logs
* Calculate total parked vehicles
* Identify peak parking usage
* Alert if parking exceeds capacity



```python
def smart_parking(capacity, logs):
    current_count = 0
    peak_usage = 0

    for log in logs:
        if log == "IN":
            current_count += 1
        elif log == "OUT":
            current_count -= 1

        # Track peak usage
        if current_count > peak_usage:
            peak_usage = current_count

    print("Currently Parked Vehicles:", current_count)

    if current_count > capacity:
        print("Parking Status: Exceeded Capacity!")
    else:
        print("Parking Status: Available")

# Sample Input
capacity = 5
logs = ["IN", "IN", "IN", "OUT", "IN", "IN", "OUT"]

smart_parking(capacity, logs)
```

# **Problem Statement 2:Online Food Delivery Time Estimator**
## **Create a function that estimates delivery time based on:**
*   Distance (km)
*   Weather condition
*   Traffic level











```python
def estimate_delivery_time(distance, traffic, weather):
    # Base time
    total_time = distance * 5

    # Traffic delay
    if traffic == "Low":
        total_time += 5
    elif traffic == "Medium":
        total_time += 10
    elif traffic == "High":
        total_time += 5

    # Weather delay
    if weather == "Clear":
        total_time += 0
    elif weather == "Cloudy":
        total_time += 5
    elif weather == "Rainy":
        total_time += 10

    print("Estimated Delivery Time:", total_time, "minutes")


# Sample Input
distance = 8
traffic = "High"
weather = "Rainy"

estimate_delivery_time(distance, traffic, weather)
```


# **Problem Statement 4:Cloud Server Load Classification System**
## **Create a function to classify server load based on CPU usage readings.**
## Rules:

* Average CPU < 50% → Normal

* 50%-80% → Warning

* 80% → Critical




```python
def classify_server_load(cpu_readings):

    # Calculate average CPU usage
    avg_cpu = sum(cpu_readings) / len(cpu_readings)

    # Determine server status
    if avg_cpu < 50:
        status = "Normal"
    elif avg_cpu <= 80:
        status = "Warning"
    else:
        status = "Critical"

    # Print results
    print("Average CPU Load:", round(avg_cpu), "%")
    print("Server Status:", status)


# Sample Input
cpu_readings = [45, 60, 70, 85, 90]

# Function call
classify_server_load(cpu_readings)
```

# **Problem Statement 5:Smart Classroom Resource Usage Monitor**
## **Design a function that tracks usage of classroom resources (projector, AC, lights) and identifies overuse patterns.**
**Real-Time Use**
* Smart classrooms
* Energy optimization systems









```python
def resource_monitor(resources):

    overused = []

    for resource, hours in resources.items():
        if hours > 8:
            overused.append(resource)

    if overused:
        print("Overused Resources:", ", ".join(overused))
        print("Energy Alert: Yes")
    else:
        print("Overused Resources: None")
        print("Energy Alert: No")


# Sample Input
resources = {
    "Projector": 6,
    "AC": 9,
    "Lights": 4
}

# Function call
resource_monitor(resources)
```

# **Problem Statement 6: Online Event Registration Capacity Controller**
## **Create a function that manages event registrations by:**

* Tracking registrations
* Preventing overbooking
* Triggering waitlist mode




```python
def event_capacity_controller(capacity, registrations):

    if registrations <= capacity:
        confirmed = registrations
        waitlist = 0
        status = "Open"
    else:
        confirmed = capacity
        waitlist = registrations - capacity
        status = "Closed"

    print("Confirmed Registrations:", confirmed)
    print("Waitlisted Users:", waitlist)
    print("Registration Status:", status)


# Sample Input
capacity = 100
registrations = 105

# Function call
event_capacity_controller(capacity, registrations)
```
